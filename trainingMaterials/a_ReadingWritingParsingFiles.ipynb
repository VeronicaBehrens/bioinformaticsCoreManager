## Reading files

### Reading an entire file as a single string

```python
INPUT_FILENAME = 'example.txt'  # store the name of the input file as a string

inFile = open(INPUT_FILENAME, 'r')   # open the file for reading (the 'r' indicates reading mode)

contents = inFile.read()  # store all the contents of the file in a variable

inFile.close()  # close the file

print(contents)
```

```python
# What data type is stored in the variable contents?

type(contents)

# So we can do whatever we'd normally do to a string!

print('8' in contents)
print('32' in contents)
#print(8 in contents)

contents = contents.replace('20', 'hello')
print(contents)
```

### Looping over a file line by line
In practice, this is used much more than reading in an entire file at once as a single string!

#### Option 1

```python
INPUT_FILENAME = 'example.txt'  # store the name of the input file as a string

inFile = open(INPUT_FILENAME, 'r')   # open the file for reading (the 'r' indicates reading mode)

for line in inFile:  # loop over every line in the file, temporarily storing each line as a string
    print(line)      # do something to each line

inFile.close()  # close the file
```

#### Option 2
This is the option Veronica usually uses.

```python
INPUT_FILENAME = 'example.txt'  # store the name of the input file as a string

with open(INPUT_FILENAME, 'r') as inFile:   # open the file for reading (the 'r' indicates reading mode)
    for line in inFile:  # loop over every line in the file, temporarily storing each line as a string
        print(line)  # do something to each line

# No need to close the file, it only stays open during the for loop!
```

## Writing files

### A very simple example

#### Option 1

```python
OUTPUT_FILENAME = 'exampleOutput.txt'  #  store the name of the output file as a string

myString1 = 'I like dogs!'
myString2 = 'I like cats!'
myString3 = 'I like horses!'

outFile = open(OUTPUT_FILENAME, 'w')  # open the file for writing (the 'w' indicates writing mode)

# You can write to a file using .write() or print()
outFile.write(myString1 + '\n')
outFile.write(myString2 + '\n')
#outFile.write(myString3 + '\n')
print(myString3, file=outFile)

outFile.close()  # close the file
```

#### Option 2
This is the option Veronica uses in practice.

```python
OUTPUT_FILENAME = 'exampleOutput.txt'  # store the name of the outputfile as a string

myString1 = 'I like dogs!'
myString2 = 'I like cats!'
myString3 = 'I like horses!'

with open(OUTPUT_FILENAME, 'w') as outFile:  # open the file for writing (the 'w' indicates writing mode)
    # You can write to a file using .write() or print()
    outFile.write(myString1 + '---')
    outFile.write(myString2 + '---')
    print(myString3, file=outFile, end='---')

# No need to close the file, it only stays open during the for loop!
```

## Parsing files

#### A very simple example to get us started.
Let's add <code>'hello'</code> to the end of each line in <code>'example.txt'</code>.

```python
INPUT_FILENAME = 'example.txt'  # store the name of the input file as a string

with open(INPUT_FILENAME, 'r') as inFile:  # open the file for reading (the 'r' indicates reading mode)
    for line in inFile:
        line = line.strip().split('\t')  # strip off the white space and make the line a list instead of a string
        line.append('hello')  # add hello to the end of each list
        print('\t'.join(line))  # make the output tab delimited
```

*Note:*

<code>line.strip()</code> removes white space from both the left and right sides of a line<br>
<code>line.rstrip()</code> removes white space from only the right side of a line<br>
<code>line.lstrip()</code> removes white space from only the left side of a line<br><br>

<code>line.strip('\n')</code> removes only new lines (not all white space) from the left and right sides of a line<br>
<code>line.strip('\t')</code> removes only tabs (not all white space) from the left and right sides of a line

#### Typical workflow when reading, writing, & parsing files
- Open a file to read
- Do some type of manipulation to each line
- Do one or more of the following:
    - Use a print statement to report information
    - Write the information to an output file
    - Store the information in a data structure and then do something with that data structure
