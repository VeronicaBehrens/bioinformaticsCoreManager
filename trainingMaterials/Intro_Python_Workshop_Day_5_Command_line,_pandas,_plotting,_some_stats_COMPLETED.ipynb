###Exercise 1:  pandas, plotting, and correlation coefficients (mouse tumor experiment)

The file <code>mouseTumorDrugExperiment.txt</code> contains the results of an experiment where wildtype mice and mutant mice were treated with a cancer drug.  The following information is in tab-delimited format for each mouse:
>- ID number
>- genotype (WT or Mut)
>- sex (M or F)
>- treated with the drug or not (Y or N)
>- body weight
>- life span (i.e. the day the mouse died, monitoring occured for 30 days)
>- tumor weight

######**1.a.** Parse <code>mouseTumorDrugExperiment.txt</code> and store each category of information in a separate list.<br><br>i.e. Store all the genotypes in a list, all the body weights in a separate list, all the tumor weights in a separate list, etc.

```
# YOUR SOLUTION HERE
mouseTumorFileName = 'mouseTumorDrugExperiment.txt'

mouseID_list = []
geno_list = []
sex_list = []
drug_list = []
bodyWeight_list = []
lifeSpan_list = []
tumorWeight_list = []

with open(mouseTumorFileName, 'r') as inFile:
    inFile.readline() # skip the header

    for line in inFile:
        mouseID, geno, sex, drug, bodyWeight, lifeSpan, tumorWeight = line.strip().split('\t')
        mouseID = int(mouseID)
        bodyWeight = float(bodyWeight)
        lifeSpan = float(lifeSpan)
        tumorWeight = float(tumorWeight)

        # save all info from the file in a separate list
        mouseID_list.append(mouseID)
        geno_list.append(geno)
        sex_list.append(sex)
        drug_list.append(drug)
        bodyWeight_list.append(bodyWeight)
        lifeSpan_list.append(lifeSpan)
        tumorWeight_list.append(tumorWeight)

print(lifeSpan_list)
```

######**1.b.** Use matplotlib to make scatter plots and find correlation coefficients with `spearmanr()` / `pearsonr()` from `scipy.stats` to answer these questions!
>- Is there a strong correlation between body weight and tumor weight?<br>
>- Is there a strong correlation between life span and tumor weight?<br>

Make sure to include axis labels and a titles on your plots!<br>

BONUS:  Can you make your two plots into a single figure with two sub-panels?<br>

```
# YOUR SOLUTION HERE

# Find the correlation coefficient and associated p-value for body weight v. tumor weight
#spearmanr()
#pearsonr()
print(spearmanr(bodyWeight_list, tumorWeight_list))
print(pearsonr(bodyWeight_list, tumorWeight_list))


# Find the correlation coefficient and associated p-value for life span v. tumor weight
#spearmanr()
#pearsonr()
print(spearmanr(lifeSpan_list, tumorWeight_list))
print(pearsonr(lifeSpan_list, tumorWeight_list))

################################################################

# Scatter plot of body weight v. tumor weight with matplotlib
plt.scatter(tumorWeight_list, bodyWeight_list, c='r')
plt.xlabel('tumor weight (g)')
plt.ylabel('body weight (g)')
plt.title('Correlation: Body weight & Tumor weight')
plt.show()

# Scatter plot of life span v. tumor weight with matplotlib
plt.scatter(tumorWeight_list, lifeSpan_list, c='g')
plt.xlabel('tumor weight (g)')
plt.ylabel('life span (days)')
plt.title('Correlation: Life span & Tumor weight')
plt.show()
```

######**4.e.** The original data is 4-dimensional (i.e. it has 4 features - sepal length, sepal width, petal length, and petal width). <br>
Here, we project the original 4-dimensional data into 2 dimensions.<br>
There is no longer a particular meaning for each principal component.<br>
The principal components are just the two main dimensions of variation.

```
# YOUR SOLUTION HERE

num_componenets = 2 # specify we should use 2 principal components. change to 3 for part 4.i.

pca = PCA(n_components=num_componenets)

principalComponents = pca.fit_transform(features_df)

principal_df_labels = ['principal component ' + str(num) \
                       for num in range(1, num_componenets+1)]

principal_df = pd.DataFrame(data = principalComponents,
                            columns = principal_df_labels)

principal_df.head()
```

######**4.f.** Here, we find the percent variance explained by each pricipal componenet.<br>
The explained variance tells you how much information (variance) can be attributed to each principal component. This is important becasue we lose some of the variance (information) when we project 4-dimentional information to 2-dimensions. By using the attribute <code>explained_variance_ratio_</code>, we see that the first principal component contains 72.77% of the variance and the second principal component contains 23.03% of the variance. Together, the two components contain 95.80% of the information.

```
# YOUR SOLUTION HERE

variance_explained_list = list(pca.explained_variance_ratio_)
print(variance_explained_list)
```

### Exercise 6:  comprehensions and <code>map()</code>

Find the complement of <code>seq</code> using:
>**A.** a for loop<br>
>**B.** a comprehension<br>
>**C.** the map() function<br>

Your output should be a string.

```
# YOUR SOLUTION: Using a for loop

seq = 'ATCGATCAGCATCTATATTCGGTC'

compDict = {'A':'T', 'T':'A', 'G':'C', 'C':'G'}

seq_comp = ''

for base in seq:
    base_comp = compDict[base] # look up the complement of each base in the dictionary
    seq_comp += base_comp # concatonate each complemented base to the existing sequence string

print(seq_comp)
```

```
# YOUR SOLUTION: Using a comprehension

seq = 'ATCGATCAGCATCTATATTCGGTC'

seq_comp = ''.join([compDict[base] for base in seq])

print(seq_comp)
```

```
# YOUR SOLUTION: Using map()

seq = 'ATCGATCAGCATCTATATTCGGTC'

seq_comp = ''.join(list(map(compDict.get, seq)))

print(seq_comp)
```

### Exercise 7:  loops and functions

Write a function called <code>find_score()</code> that scores how similar two DNA sequences are to each other.
>- matches are worth +3<br>
>- mismatches are worth -2

Your function should return (as a tuple):
>1. the score
>2. the number of matches
>3. the number of mismatches

For example, these two DNA sequences would have a score of 10, <br>
with 4 matches (each worth +3) and 1 mismatch (worth -2)<br><br>
<code>ACATC</code><br>
<code>|||x|</code><br>
<code>ACACC</code>

Use your function on the following pairs:
>1. <code>seq1</code> and <code>seq2</code>
>2. <code>seq2</code> and <code>seq3</code>
>3. <code>seq1</code> and <code>seq3</code>

```
seq1 = 'TCCAATCGGGCTAATTCGGCTAGAGTCGGCTAGAGAGGGAGA'
seq2 = 'TCCAGTCGGGCTTATACGGCTACTGTCCGCTAGTGACCCAGA'
seq3 = 'TGCAATGGGGCTAATACGGCAAGAGTCCCCTAGAGAGGGAGT'
```

```
# YOUR FUNCTION HERE

def find_score(seq1, seq2):
    score = 0
    numMatches = 0
    numMismatches = 0

    for nucleotide1, nucleotide2 in zip(seq1, seq2):
        if nucleotide1 == nucleotide2:
            score += 3
            numMatches += 1
        else:
            score -= 2
            numMismatches += 1

    return score, numMatches, numMismatches
```

```
# USE THE FUNCTION

score, matches, mismatches = find_score(seq1, seq2)
print(score, matches, mismatches)

score, matches, mismatches = find_score(seq1, seq3)
print(score, matches, mismatches)

score, matches, mismatches = find_score(seq2, seq3)
print(score, matches, mismatches)
```
