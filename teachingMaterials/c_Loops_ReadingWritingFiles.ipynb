## for loops
#### iterating over strings, lists, tuples, sets, dictionaries

```python
# Iterating over a string

myString = 'stuvwxyz'

for item in myString:
    print(item)
```

```python
# Iterating over a list

myList = [1, 2, 3, 4, 5, 6]

for item in myList:
    print(item, item*5)
```

```python
# Iterating over a dictionary

myDict = {'a': 1, 'b':2, 'c':3, 'd':4}

# Option 1
for key, value in myDict.items():
    print(key, value, value*5)
    
print()

# Option 2
for key in myDict:
    value = myDict[key]
    print(key, value, value*5)
```

## range(), zip(), and emumerate(): Useful tools for loops

#### range()

```python
for item in range(0, 5):
    print(item)
```

#### zip()

```python
# Iterate over multiple data structures at the same time

list1 = [1, 2, 3, 4] 
list2 = [10, 11, 12, 13]
string3 = 'abcdefg'

for item1, item2, item3 in zip(list1, list2, string3):
    print(item1, item2, item3)
```

#### enumerate()

```python
# Iterate over a data strucutre with a built-in counter

myString = 'stuvwxyz'

for i, item in enumerate(myString):
    print(i, item)
    

# enumerate does exactly the same thing as this!
#i = 0 
#for item in myString:
#    print(i, item)
#    i += 1
```

## Reading and Writing Files

In the following examples, the script adds <code>'hello'</code> to the end of each line in <code>'example.txt'</code>.

#### Workflow 1
- Open a file to read
- Do some type of manipulation to each line
- Use a print statement to report information

```python
INPUT_FILENAME = 'example.txt'

with open(INPUT_FILENAME, 'r') as inFile:
    for line in inFile:
        #print(line)
        line = line.strip().split('\t')
        #print(line)
        line.append('hello')
        print('\t'.join(line))
```

#### Workflow 2
- Open a file to read AND a file to write
- Do some type of manipulation to each line
- Write information to a new file

```python
# Option 1

INPUT_FILENAME = 'example.txt'
OUTPUT_FILENAME = 'example_hello.txt'

with open(INPUT_FILENAME, 'r') as inFile, open(OUTPUT_FILENAME, 'w') as outFile:
    for line in inFile:
        line = line.strip().split('\t')
        line.append('hello')
        outFile.write('\t'.join(line) + '\n')
```

```python
# Option 2

INPUT_FILENAME = 'example.txt'
OUTPUT_FILENAME = 'example_hello.txt'

with open(INPUT_FILENAME, 'r') as inFile:
    with open(OUTPUT_FILENAME, 'w') as outFile:
        for line in inFile:
            line = line.strip().split('\t')
            line.append('hello')
            outFile.write('\t'.join(line) + '\n')
```

#### Workflow 3
- Open a file to read
- Store information in some type of data structure
- Use that data structure to do some type of manipulation

```python
INPUT_FILENAME = 'example.txt'

lineList = []

with open(INPUT_FILENAME, 'r') as inFile:
    for line in inFile:
        line = line.strip().split('\t')
        lineList.append(line)
        
#print(lineList)

for item in lineList:
    item.append('hello')
    print('\t'.join(item))
```

# ('{:}').format()

**'{**  <code>positionalInformation</code>  **:**  <code>numberOfSpacesBefore</code>  **.**  <code>numberOfDigitsAfterDecimal</code>  <code>dataTypeAbbreviation</code>  **}'**<br><br>


For example:  <code>'{2:5.3f}'</code> means:
>- use variable in position 3 within the {''}.format() parentheses
>- add 5 spaces before
>- truncate to 3 digits
>- the data type is a float

A few of the most useful data type abbreviations:
>- <code>d</code>: integer
>- <code>e</code>: scientific notation
>- <code>f</code>: float
>- <code>s</code>: string

```python
print('Here are three variables: {} {} {}'.format(5, 'abc', 4.531))
print('Here are three variables: {1} {2} {0}'.format(5, 'abc', 4.531))
print('Here are four variables: {1} {2} {0} {1}'.format(5, 'abc', 4.531))
```

```python
print('Here are three variables: {} {} {:.2f}'.format(5, 'abc', 4.531))
print('Here are three variables: {} {} {:.1f}'.format(5, 'abc', 4.531))
print('Here are three variables: {:} {} {:15.1f}'.format(5, 'abc', 4.531))
print('Here are three variables: {:4d} {} {:.1e}'.format(5, 'abc', 4.531))
```

### Exercise 3

The file <code>sequences.tsv</code> contains the following information:
>- chromosome
>- starting coordinate
>- the DNA sequence

Use <code>sequences.tsv</code> to generate a new tab-delimited file (called <code>sequences_GC_content.tsv</code>)
that contains only sequenes with GC content below 60% and reports the following information:
>- chromosome
>- starting coordinate
>- ending coordinate
>- the DNA sequence
>- GC content (rounded to 2 decimal places)

```python
# YOUR SOLUTION HERE

inFileName = 'sequences.tsv'
outFileName = 'sequences_GC_content.tsv'
THRESHOLD = 0.6

# open the file (both the input and the output file)
with open(inFileName, 'r') as inFile, open(outFileName, 'w') as outFile:
    
    # begin a for loop to go line by line through the file
    for line in inFile:
        # manipulate the line to remove line breaks and split up the line based on tabs
            # store the chromosome, start coord, and sequence in variables
        chrom, start, seq = line.strip().split('\t')
        seq = seq.upper()
        #print(seq)
        
        # calculate the end coordinate
            # find the length of the DNA sequence
            # arithmetic to get the end coord (and store in variable)
        length = len(seq)
        end = length + int(start)
        #print(end)

        # calculate GC content (could use myString.count() or a loop)
            # calculate the number of G's
            # calculate the number of C's
            # calculate the total number of bases (length of DNA sequence)
            # arithmetic to get the GC proportion
        g_count = seq.count('G')
        c_count = seq.count('C')
        gc_content = (g_count + c_count) / length
        #print(gc_content)
         
        #Check if the GC content is less than 0.6
            # if it is, write the information to the output file
        if gc_content < THRESHOLD:
            #print(chrom, start, end, seq, round(gc_content, 2))
            toPrint = [chrom, start, str(end), seq, str(round(gc_content, 2))]
            toPrint = '\t'.join(toPrint)
            outFile.write(toPrint + '\n')
```
