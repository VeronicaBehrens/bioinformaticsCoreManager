#### REVIEW:  A very simple parsing example to get us started.
Let's add <code>'hello'</code> to the end of each line in <code>'example.txt'</code>.

````python
INPUT_FILENAME = 'example.txt'  # store the name of the input file as a string

with open(INPUT_FILENAME, 'r') as inFile:  # open the file for reading (the 'r' indicates reading mode)
    for line in inFile:
        line = line.strip().split('\t')  # strip off the white space and make the line a list instead of a string
        line.append('hello')  # add hello to the end of each list
        print('\t'.join(line))  # make the output tab delimited
````

*Note:*

<code>line.strip()</code> removes white space from both the left and right sides of a line<br>
<code>line.rstrip()</code> removes white space from only the right side of a line<br>
<code>line.lstrip()</code> removes white space from only the left side of a line<br><br>

<code>line.strip('\n')</code> removes only new lines (not all white space) from the left and right sides of a line<br>
<code>line.strip('\t')</code> removes only tabs (not all white space) from the left and right sides of a line

#### Typical workflow when reading, writing, & parsing files
- Open a file to read
- Do some type of manipulation to each line
- Do one or more of the following:
    - Use a print statement to report information
    - Write the information to an output file
    - Store the information in a data structure and then do something with that data structure

Let's add together all the numbers on each line of <code>example.txt</code> and append the sum to the end of each line.

````python
INPUT_FILENAME = 'example.txt'
OUTPUT_FILENAME = 'workflowOutput.csv'

# ADD CODE HERE

with open(INPUT_FILENAME, 'r') as inFile, open(OUTPUT_FILENAME, 'w') as outFile:
  for line in inFile:
    line = line.strip().split('\t')

    # convert all items in line (which is now a list) to integers
    myIntList = []
    for item in line:
      item = int(item)
      myIntList.append(item)

    # find the sum of the integers in the line and append the sum to the list
    currentSum = sum(myIntList)
    myIntList.append(currentSum)

    # convert all item in myIntList to strings
    myStrList = []
    for item in myIntList:
      item = str(item)
      myStrList.append(item)

    print(','.join(myStrList)) # print the result to the screen
    print(','.join(myStrList), file=outFile) # write the result to a file
````

## Comprehensions and `map()`:  Condensing for loops

**EXAMPLE:  Three ways to convert a list of strings to a list of integers**

##### **1. Using a for loop**

````python
myList = ['1', '2', '3', '4']

myIntList = []

for num in myList:
    num = int(num)
    myIntList.append(num)

print(myIntList)
````

##### **2. Using a comprehension**

<b>[</b> do this expression <b>for</b> item <b>in</b>  myList <b>if</b> conditional <b>]</b>

```
for item in myList:
  if conditional:
    expression
```

````python
myList = ['1', '2', '3', '4']

myList = [int(item) for item in myList]

print(myList)
````

````python
# Here's how to convert to an integer AND multiply by 3 for each item in the list
#    ONLY IF the original number is greater than 2!

myList = ['1', '2', '3', '4']

myList = [int(x)*3 for x in myList if int(x) > 2]

print(myList)
````

````python
# You can do dictionary comprehensions, too!
# Here's how to raise each value in a dictionary to the power of 2.

myDict = {'a': 1, 'b': 2, 'c': 3, 'd': 4, 'e': 5}

myDict = {key:value**2 for key, value in myDict.items()}

print(myDict)
````

##### **3. Using the <code>map()</code> function**
map(<b>do</b> *this function*, <b>to each item</b> in *this list*)

````python
myList = ['1', '2', '3', '4']

myList = list(map(int, myList))
print(myList)
````

### Exercise 2

The file <code>sequences.tsv</code> contains the following tab-delimited information:
>- chromosome
>- starting coordinate
>- the DNA sequence

The file <code>enzymes.csv</code> contains the following comma-delimited information:
>- restriction enzyme name
>- rescriction enzyme recognition sequence

Use <code>sequences.tsv</code> to generate a new tab-delimited file (called <code>sequencesResults.tsv</code>)
that contains ONLY SEQUENCES ON CHR2 with GC CONTENT BELOW 60% and reports the following information:
>- chromosome
>- starting coordinate
>- ending coordinate
>- the DNA sequence
>- which enzymes will cut DNA sequence (comma-delmited if more than one, for example EcoRI,NotI)
>- GC content of the DNA sequence (rounded to 2 decimal places)

````python
enzymesFileName = 'enzymes.csv'
sequencesFileName = 'sequences.tsv'
outFileName = 'sequencesResults.tsv'

GC_THRESHOLD = 60

# Start by storing the information in enzymes.csv in a dictionary
enzymesDict = {}
with open(enzymesFileName, 'r') as inFile:
  for line in inFile:
    enzName, recSite = line.strip().split(',')
    #enzName = line[0]
    #recSite = line[1]
    enzymesDict[enzName] = recSite

#print(enzymesDict)

# Loop through sequences.tsv line by line (and open our output file)
with open(sequencesFileName, 'r') as inFile, open(outFileName, 'w') as outFile:
  for line in inFile:
    chrom, start, seq = line.strip().split('\t') # store chromosome, start coordinate, and sequence in variables
    start = int(start) # make the start coordinate and integer so we can do math with it

    # Start by limiting the results to sequences on chr2
    if chrom == 'chr2':

      # Then, find the end coordinate of the DNA sequence
      length = len(seq)
      end = start + length

      # Next, figure out which enzymes cut the DNA sequence
      enzymesThatCutList = []
      for enzName, recSite in enzymesDict.items(): # loop through our dictionary of enzymes and recognition sites
        if recSite in seq.upper(): # check if the recognition site is in the DNA sequence
          enzymesThatCutList.append(enzName) # store the name of the enzymes that will cut in a list

      # Finally, calculate the GC content of the sequence
      g_count = seq.upper().count('G') # count the number of Gs
      c_count = seq.upper().count('C') # count the number of Cs
      gc_content = ((g_count + c_count) / length) * 100 # calculate the GC content
      gc_content = round(gc_content, 2) # round the GC content to 2 decimal places

      # ONLY IF THE GC CONTENT IS LESS THAN 60%, write the results to sequencesResults.tsv in the format we want
      if gc_content < GC_THRESHOLD:
          enzymesThatCutList = ','.join(enzymesThatCutList)
          myListForPrinting = [chrom, str(start), str(end), seq, enzymesThatCutList, str(gc_content)]
          print('\t'.join(myListForPrinting))
          print('\t'.join(myListForPrinting), file=outFile)
````

##### **There are different ways to import libraries**

````python
# import the math library
# find the sqrt of 9
import math
print(math.sqrt(9))

# import the math library, but call it m
# find the sqrt of 9
import math as m
print(m.sqrt(9))

# import ONLY the sqrt function from math
# find the sqrt of 9
from math import sqrt
print(sqrt(9))

# import ONLY the sqrt function from math, but call it s
# find the sqrt of 9
from math import sqrt as s
print(s(9))
````
